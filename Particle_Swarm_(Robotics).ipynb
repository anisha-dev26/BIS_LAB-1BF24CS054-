```python

import numpy as np

np.random.seed(42)

# Environment
START = np.array([0.0, 0.0])
GOAL = np.array([10.0, 10.0])

# Obstacles: (x, y, radius)
OBSTACLES = [
    (4.0, 4.0, 1.5),
    (6.0, 6.0, 1.5)
]

NUM_PARTICLES = 60
GENERATIONS = 100
NUM_WAYPOINTS = 4

W = 0.7
C1 = 1.5
C2 = 1.5
PENALTY = 1000


def create_path(position):
    waypoints = position.reshape(NUM_WAYPOINTS, 2)
    return np.vstack([START, waypoints, GOAL])


def path_length(path):
    return sum(
        np.linalg.norm(path[i + 1] - path[i])
        for i in range(len(path) - 1)
    )


def collision_penalty(path):
    penalty = 0

    for i in range(len(path) - 1):
        a, b = path[i], path[i + 1]

        for t in np.linspace(0, 1, 100):
            point = a + t * (b - a)

            for ox, oy, radius in OBSTACLES:
                center = np.array([ox, oy])

                if np.linalg.norm(point - center) < radius:
                    penalty += 1

    return penalty


def fitness(position):
    path = create_path(position)
    distance = path_length(path)
    collisions = collision_penalty(path)

    return distance + PENALTY * collisions


# Initialize particles
dimensions = NUM_WAYPOINTS * 2

particles = np.random.uniform(
    0, 10, (NUM_PARTICLES, dimensions)
)
velocities = np.zeros_like(particles)

personal_best = particles.copy()
personal_fitness = np.array([
    fitness(p) for p in particles
])

best_index = np.argmin(personal_fitness)
global_best = personal_best[best_index].copy()
global_fitness = personal_fitness[best_index]

# PSO iterations
for generation in range(GENERATIONS):

    for i in range(NUM_PARTICLES):
        r1 = np.random.rand(dimensions)
        r2 = np.random.rand(dimensions)

        velocities[i] = (
            W * velocities[i]
            + C1 * r1 * (personal_best[i] - particles[i])
            + C2 * r2 * (global_best - particles[i])
        )

        particles[i] += velocities[i]
        particles[i] = np.clip(particles[i], 0, 10)

        current_fitness = fitness(particles[i])

        if current_fitness < personal_fitness[i]:
            personal_best[i] = particles[i].copy()
            personal_fitness[i] = current_fitness

        if current_fitness < global_fitness:
            global_best = particles[i].copy()
            global_fitness = current_fitness

    print(
        f"Generation {generation + 1}: "
        f"Best Fitness = {global_fitness:.4f}"
    )


# Final solution
best_path = create_path(global_best)
final_distance = path_length(best_path)
final_collisions = collision_penalty(best_path)

print("\n===== FINAL SOLUTION =====")
print("Start Position:", START)

for i, point in enumerate(best_path[1:-1], start=1):
    print(f"Waypoint {i}: ({point[0]:.2f}, {point[1]:.2f})")

print("Goal Position:", GOAL)
print(f"Total Path Length: {final_distance:.2f}")
print(f"Collision Penalty: {final_collisions}")

if final_collisions == 0:
    print("Result: Collision-free path found.")
else:
    print("Result: Path intersects an obstacle.")
```
